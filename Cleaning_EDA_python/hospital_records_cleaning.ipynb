```python
def basic_checks(df):
    print(f"Shape of data(Rows * Columns)\n{df.shape}")
    print("-"*100)
    print(f"Datatypes:\n {df.info()}")
    print("-"*100)
    print(f"Numerical stats:\n {display(df.describe(include=['int','float']).T)}")
    print("-"*100)
    print(f"Categorical stats:\n {display(df.describe(include=['str']).T)})")
    print("-"*100)
    print(f"Missing values:\n {(df.isnull().sum())/(len(df))}")
    print("-"*100)
    print(f"Duplicated Values:\n{df.duplicated().sum()}")
    print("-"*100)
```

```python
basic_checks(hosp_df)
```

```python
missing_cols = hosp_df[['Age','Gender','Diagnosis']]

print(missing_cols)
```

```python
num_nulls = hosp_df['Age']
cat_nulls = hosp_df[['Gender','Diagnosis']]
```

```python
def quality_fix(df):
    df['Age'] = df['Age'].fillna(df['Age'].mean())
    df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
    df['Diagnosis'] = df['Diagnosis'].fillna('No diagnosis given')
    print(df.isnull().sum())
      
```

```python
quality_fix(hosp_df)
```
